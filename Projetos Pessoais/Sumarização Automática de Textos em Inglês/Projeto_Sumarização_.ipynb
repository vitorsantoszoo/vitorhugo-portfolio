## ⚙️ Como o projeto funciona

   Utilizamos o **DistilBART CNN-12-6**, uma versão compacta do BART, otimizada para velocidade e boa qualidade de resumo.

4. **Geração do resumo**  
   Em seguida, o modelo devolve uma versão curta e coesa do texto, seguindo os padrões linguísticos aprendidos durante seu treinamento.

5. **Exibição formatada**

# 🟦 CÉLULA 2 — Importação das bibliotecas necessárias
Aqui importamos os módulos usados no projeto, incluindo:
- pipeline (Transformers)
- AutoTokenizer
- AutoModelForSeq2SeqLM (arquitetura encoder-decoder)


```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM
```

# 🟦 CÉLULA 3 — Carregar modelo T5 pré-treinado para sumarização
Nesta etapa, carregamos o modelo T5-base-portuguese, treinado pela Unicamp (UNICAMP-DL).
Ele é capaz de realizar tarefas de sumarização, tradução e transformação de texto em PT-BR.


```python
summarizer = pipeline(
    task="summarization",
    model="sshleifer/distilbart-cnn-12-6"
)
```

# 🟦 CÉLULA 4 — Definir texto de entrada
Aqui colocamos um texto comum para ser resumido.
Você pode alterar o texto para qualquer artigo, notícia, parágrafo ou informação que desejar.


```python
text = """
Artificial intelligence has become one of the most influential technologies today, driving significant advances in areas such as medicine, industry, education, and security.
Modern models, such as Transformers, allow machines to understand and generate natural language with impressive accuracy, enabling applications like virtual assistants, automatic translation,
text summarization, and sentiment analysis. With the rapid growth of this technology, experts discuss not only its benefits but also the associated risks, including ethical issues, privacy concerns, and its impact on the job market.
"""
text
```

# 🟦 CÉLULA 5 — Gerar resumo automático com T5
Aqui utilizamos o pipeline de sumarização para produzir um resumo conciso do texto informado.
Podemos ajustar parâmetros como:
- max_length: tamanho máximo do resumo
- min_length: tamanho mínimo do resumo
- do_sample: usar ou não amostragem


```python
resumo = summarizer(
    texto,
    max_length=60,
    min_length=25,
    do_sample=True
)

resumo[0]["summary_text"]
```

# 🟦 CÉLULA 6 — Exibir texto original e resumo de forma legível

Aqui exibimos o texto original e seu resumo, para facilitar a análise do resultado.


```python
# Exibir texto original formatado
print("========== TEXTO ORIGINAL (EN) ==========\n")
print(text.strip())
print("\n" + "="*60 + "\n")

# Exibir resumo gerado pelo modelo de forma legível
print("========== RESUMO GERADO PELO MODELO (EN) ==========\n")

summary_text = resumo[0]["summary_text"]

# Quebrar linhas automaticamente a cada 100 caracteres (ajuste se quiser)
import textwrap
wrapped_summary = textwrap.fill(summary_text, width=100)

print(wrapped_summary)
print("\n" + "="*60)
```

### 💬 Comentário sobre o desempenho do modelo

O modelo conseguiu identificar os principais avanços descritos no texto e sintetizou bem a ideia central sobre a evolução dos modelos de linguagem e sua capacidade de interpretar e gerar conteúdo.



# 🟦 CÉLULA 7 — Criar função para sumarizar qualquer texto usando DistilBART
Esta função permite testar rapidamente novos textos copiando e colando no Colab,
sem precisar reescrever código.


```python
def resumir(texto, maxlen=60, minlen=25):
    resumo = summarizer(
        texto,
        max_length=maxlen,
        min_length=minlen,
        clean_up_tokenization_spaces=True
    )
    return resumo[0]["summary_text"]
```

# 🟦 CÉLULA 8 — Exibir texto original e resumo (EN) de forma legível
Aqui você pode testar qualquer texto adicional,
como trechos de artigos, notícias, Wikipedia ou descrições técnicas.


```python
novo_texto = """
Recent advances in language models have enabled the creation of systems capable of interpreting complex instructions, answering contextualized questions, and generating content naturally.
These models use billions of parameters and are trained on large amounts of textual data, allowing them to recognize sophisticated patterns in language.

"""

print(resumir(novo_texto))
```

```python
# Exibir texto original da CÉLULA 8
print("========== TEXTO ORIGINAL (EN) — CÉLULA 8 ==========\n")
print(novo_texto.strip())   # <-- assegure-se de usar a variável correta
print("\n" + "="*60 + "\n")

# Exibir resumo gerado pelo modelo
print("========== RESUMO GERADO PELO MODELO (EN) — CÉLULA 8 ==========\n")

summary8 = summarizer(
    novo_texto,
    max_length=60,
    min_length=25,
    clean_up_tokenization_spaces=True
)[0]["summary_text"]

# Quebra automática de linhas — mais legível
import textwrap
wrapped_summary8 = textwrap.fill(summary8, width=100)

print(wrapped_summary8)
print("\n" + "="*60)
```
