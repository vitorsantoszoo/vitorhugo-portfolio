# 💬 AI Agent Hub — Agente de Linguagem (NLG por Regras e Templates)
Este notebook implementa o **Agente de Linguagem (Language Agent)**, responsável
por ler os resultados técnicos dos outros agentes (Visão, Preditivo, etc.)
e gerar um **relatório consolidado em linguagem natural**.

🧠 Funções principais:
1️⃣ Ler logs JSON dos outros agentes (pasta /content/data/logs/)
2️⃣ Interpretar métricas e status
3️⃣ Gerar textos automáticos e coerentes (sem modelos generativos)
4️⃣ Salvar o relatório consolidado em JSON e TXT

⚙️ Roda totalmente em CPU e usa apenas bibliotecas padrão do Python.



# 🧰 Imports e preparação
Nesta célula:
- Importamos bibliotecas padrão.
- Definimos o caminho base dos logs.



```python
import os
import json
from datetime import datetime
import glob
import uuid

# Caminho base dos logs dos agentes
LOG_DIR = "/content/data/logs"
OUTPUT_DIR = "/content/data/logs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print(f"📂 Lendo logs a partir de: {LOG_DIR}")
```

# 📦 Leitura e consolidação dos logs JSON dos agentes
O agente de linguagem procura automaticamente arquivos JSON criados
pelos outros agentes (vision_agent, predictive_agent, etc.).



```python
# Encontra todos os logs JSON disponíveis
log_files = glob.glob(os.path.join(LOG_DIR, "*.json"))
if not log_files:
    raise FileNotFoundError("❌ Nenhum log encontrado em /content/data/logs.")

print(f"🔍 {len(log_files)} arquivos encontrados.\n")

logs = []
for log_path in log_files:
    try:
        with open(log_path, "r", encoding="utf-8") as f:
            data = json.load(f)
            logs.append(data)
    except Exception as e:
        print(f"⚠️ Erro ao ler {log_path}: {e}")

print(f"✅ {len(logs)} logs carregados com sucesso.")
```

# 🧠 Interpretação das métricas
Traduz os resultados técnicos (ex: acurácia, erro, status)
em frases descritivas e humanas.



```python
def interpretar_agente(log):
    nome = log.get("agent", "Agente Desconhecido")
    modelo = log.get("model", "Modelo Desconhecido")
    status = log.get("status", None)
    accuracy = log.get("accuracy", None)
    observacao = ""

    if accuracy is not None:
        if accuracy >= 0.90:
            observacao = "apresentou excelente desempenho, com alta precisão e estabilidade."
        elif accuracy >= 0.75:
            observacao = "demonstrou bom desempenho, com pequenas variações entre as classes."
        else:
            observacao = "precisa de revisão, apresentando baixa acurácia ou inconsistência nos resultados."

    elif "erro" in log or "mse" in log:
        observacao = "realizou previsões numéricas com resultados satisfatórios e erro médio controlado."

    elif status:
        observacao = f"teve status final de {status.lower()} conforme avaliação técnica."

    else:
        observacao = "não apresentou métricas suficientes para análise detalhada."

    return f"🧩 {nome}: o modelo {modelo} {observacao}"
```

# 🧾 Geração do Relatório Consolidado
Cria o relatório completo com base nos logs interpretados.



```python
relatorio_texto = []
for log in logs:
    descricao = interpretar_agente(log)
    relatorio_texto.append(descricao)

texto_final = (
    "📊 Relatório Consolidado de IA — AI Agent Hub\n"
    "--------------------------------------------------\n"
    + "\n".join(relatorio_texto)
    + "\n--------------------------------------------------\n"
    f"📅 Data UTC: {datetime.utcnow().strftime('%Y-%m-%d %H:%M:%S')}\n"
    "💡 Análise geral: Os agentes operam de forma estável e coerente, "
    "com resultados consistentes para demonstração e evolução futura.\n"
)

print(texto_final)
```

# 💾 Salvamento do relatório consolidado
Salva a saída do Agente de Linguagem em formato JSON e TXT.



```python
output_json = {
    "agent": "language_agent",
    "timestamp_utc": datetime.utcnow().isoformat() + "Z",
    "relatorio_texto": texto_final,
    "origem_logs": log_files
}

json_path = os.path.join(OUTPUT_DIR, f"language_agent_report_{uuid.uuid4().hex[:8]}.json")
txt_path = json_path.replace(".json", ".txt")

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(output_json, f, indent=2, ensure_ascii=False)

with open(txt_path, "w", encoding="utf-8") as f:
    f.write(texto_final)

print(f"✅ Relatório consolidado salvo com sucesso:")
print(f"🧾 JSON: {json_path}")
print(f"📄 TXT: {txt_path}")
```
