# 📘 Projeto: Similaridade Semântica com Sentence-BERT (Encoder-Only)

## 🎯 Objetivo Geral
Investigar, implementar e avaliar uma aplicação prática baseada em representações contextuais
geradas por um modelo do tipo encoder, utilizando embeddings de sentenças (h_[CLS]) para medir
similaridade semântica entre textos.

---

## 💡 Motivação e Justificativa
Modelos do tipo encoder — como BERT e suas variações — representam o estado da arte em tarefas de
entendimento de linguagem natural. Eles convertem texto em vetores densos de alta dimensão,
capturando significados, relações e contexto.

A tarefa de *similaridade semântica* é uma das formas mais diretas e eficazes de demonstrar:

- Como o encoder gera vetores contextuais;
- Como medir proximidade entre sentenças usando métricas como similaridade de cosseno;
- Como embeddings podem ser utilizados para aplicações reais de NLP.

Essa escolha está em linha com as orientações da disciplina, que recomendam a criação de projetos
baseados explicitamente em representações geradas por um encoder.

---

## ⚙️ Metodologia

### **1. Escolha do Modelo Encoder**
Foi utilizado o modelo **multi-qa-MiniLM-L6-cos-v1**, uma variante compacta e eficiente do
Sentence-BERT (SBERT).  
Ele é composto apenas por um encoder e foi treinado especificamente para tarefas de similaridade
e recuperação de informações.

### **2. Geração de Representações Contextuais**
Cada sentença é processada pelo encoder, que produz um vetor (embedding) representando sua
informação semântica global.  
Esses vetores correspondem diretamente à representação contextual agregada da sentença,
similar à saída h_[CLS] dos modelos BERT-like.

### **3. Cálculo de Similaridade**
As similaridades são calculadas usando:

- **Similaridade de cosseno (cosine similarity)**  
  para comparar a proximidade entre vetores.

Isso permite verificar quais frases são semanticamente mais parecidas entre si.

### **4. Interpretação dos Resultados**
A partir da matriz de similaridade, é possível identificar:

- Sentenças que falam sobre temas semelhantes;
- Distinções sutis entre conceitos técnicos;
- Agrupamentos naturais formados pelas relações entre as frases.

Essa análise ilustra como o encoder captura relações semânticas mesmo entre textos curtos.

---

## 📈 Resultados e Discussão
Os resultados obtidos mostraram que o modelo foi capaz de:

- Aproximar frases que tratam de tópicos similares (ex.: frases sobre Transformers e modelos de linguagem);
- Diferenciar frases sobre arquiteturas diferentes (CNN, RNN, etc.);
- Criar agrupamentos coerentes, baseados apenas na informação contextual capturada pelo encoder.

A saída mais relevante — as sentenças mais semelhantes — demonstra o funcionamento prático do encoder
e como embeddings podem ser utilizados diretamente em soluções reais de NLP.

---

## 🧩 Conclusão
Este projeto demonstra uma aplicação clara e objetiva de modelos encoder-only na área de
Processamento de Linguagem Natural.  
Através da geração de embeddings e do cálculo de similaridade, foi possível observar de forma
concreta como representações contextuais são utilizadas para comparar sentenças.

A proposta cumpre integralmente o objetivo da disciplina: aplicar e analisar uma tarefa fundamentada
na arquitetura Encoder, explorando o uso de vetores h_[CLS] e demais embeddings para interpretação
de linguagem natural.

Instalamos a biblioteca `sentence-transformers`, que contém modelos encoder-only
derivados do BERT e otimizados para gerar vetores de sentenças. Esses vetores
correspondem às representações contextuais geradas pelo encoder, como h_[CLS].


```python
!pip install -q sentence-transformers
```

# 🟦 CÉLULA 2 — Importações necessárias
Importamos o modelo Sentence-BERT, a função utilitária para similaridade de cosseno,
e outras ferramentas auxiliares.


```python
from sentence_transformers import SentenceTransformer, util
import torch
```

# 🟦 CÉLULA 3 — Carregar modelo Encoder-Only (Sentence-BERT)
Neste projeto utilizamos o modelo **multi-qa-MiniLM-L6-cos-v1**.
Ele é um modelo encoder-only, rápido e de alta precisão, ideal para medir similaridade entre sentenças.


```python
model = SentenceTransformer("multi-qa-MiniLM-L6-cos-v1")
print("Modelo carregado com sucesso!")
```

# 🟦 CÉLULA 4 — Definir sentenças para análise de similaridade
Criamos um pequeno conjunto de frases relacionadas ao tema Inteligência Artificial
para gerar embeddings e medir o quanto elas são semanticamente parecidas.


```python
sentences = [
    "Transformers são modelos de deep learning baseados em mecanismos de atenção.",
    "Redes neurais convolucionais são amplamente usadas para processamento de imagens.",
    "Modelos de linguagem modernos conseguem gerar textos com grande fluidez.",
    "Redes neurais recorrentes eram utilizadas antes da popularização dos Transformers.",
    "Sistemas de reconhecimento de fala utilizam modelos treinados com grandes quantidades de dados.",
]
sentences
```

# 🟦 CÉLULA 5 — Gerar embeddings utilizando o encoder
Aqui o modelo encoder gera um vetor para cada frase. Esses vetores representam
as informações contextuais da sentença (h_[CLS]), permitindo calcular similaridade.


```python
embeddings = model.encode(sentences, convert_to_tensor=True)
embeddings
```

# 🟦 CÉLULA 6 — Calcular similaridade entre todas as frases
Utilizamos similaridade de cosseno (cosine similarity) para medir o quanto
as frases se relacionam semanticamente.


```python
similarity_matrix = util.cos_sim(embeddings, embeddings)
similarity_matrix
```

# 🟦 CÉLULA 7 — Exibir similaridades com clareza
Exibimos a matriz de similaridade com nomes das sentenças para facilitar a interpretação.


```python
import pandas as pd

df_sim = pd.DataFrame(
    similarity_matrix.cpu().numpy(),
    columns=[f"Sentença {i+1}" for i in range(len(sentences))],
    index=[f"Sentença {i+1}" for i in range(len(sentences))]
)
df_sim
```

# 🟦 CÉLULA 8 — Interpretação dos resultados
Nesta célula faremos uma interpretação manual das relações mais fortes encontradas
pelo modelo, simulando a análise humana e validando o comportamento do encoder.


```python
for i in range(len(sentences)):
    sims = similarity_matrix[i]

    # Criar uma cópia e zerar a similaridade da própria sentença
    sims_no_self = sims.clone()
    sims_no_self[i] = -1  # assim ela nunca será escolhida

    # Encontrar o índice da frase mais similar (excluindo ela mesma)
    most_similar_idx = torch.argmax(sims_no_self).item()

    print(f"\nFrase {i+1}: \"{sentences[i]}\"")
    print(f"Mais similar: \"{sentences[most_similar_idx]}\"")
    print(f"Similaridade: {sims_no_self[most_similar_idx]:.4f}")
```

# 🟦 CÉLULA 9 — Conclusão
O modelo Encoder-Only (SBERT) conseguiu capturar corretamente relações semânticas:
- Frases sobre Transformers tendem a se aproximar.
- Frases sobre redes neurais possuem maior similaridade entre si.
- Frases sobre aplicações práticas (fala, texto) formam outro subgrupo.

Essa análise demonstra, na prática, como representações contextuais (h_[CLS])
podem ser utilizadas para medir similaridade entre textos.
