```python
#1. Elaborar um código em Python capaz de gerar um gráfico de barras com os 10 países melhor posicionados em um ano específico, escolhido pelo grupo.

import pandas as pd
import matplotlib.pyplot as plt

file_path = "fifa_ranking-2024-04-04.csv"
df = pd.read_csv(file_path)


df['rank_date'] = pd.to_datetime(df['rank_date'])
df_filtered = df[(df['rank_date'].dt.year >= 2023) & (df['rank_date'].dt.year <= 2024)]

top_countries = (
    df_filtered.groupby("country_full")["rank"]
    .min()
    .sort_values()
    .head(10)
)

plt.figure(figsize=(10, 6))
top_countries.plot(kind="bar", color="green")
plt.title("Top 10 Países Melhor Posicionados (2023-2024)", fontsize=14)
plt.xlabel("País", fontsize=12)
plt.ylabel("Melhor Posição no Ranking", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.gca().invert_xaxis()
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```
