Atividade 4 : Pré-processamento inicial de um texto em Linguagem Python

1 - Escreva um pequeno trecho de código em linguagem Python capaz de carregar o arquivo texto  chamado "Texto_Homem_Lua.txt"   no ambiente Colab e em seguida transformar todos os caracteres do texto para minúsculos, além de retirar os números e as pontuações.

```python
import string
import re

# ==============================================================================
# TEXTO EMBUTIDO
# ==============================================================================
TEXTO_ORIGINAL = """A chegada do homem à Lua, em 20 de julho de 1969, constitui um dos maiores feitos científicos do século XX.Em 20 de julho de 1969, dois astronautas americanos, Neil Armstrong e Buzz Aldrin, se tornaram os primeiros seres humanos a por os pés em solo lunar. Um terceiro, Michael Collins,
ficou em órbita dando suporte aos companheiros.
Este feito só foi possível devido a um pesado investimento técnico-científico de 22 bilhões de dólares que envolveu mais de cem mil pessoas!!! Igualmente, na década de 60, as duas potências mundiais, Estados Unidos e União Soviética, usavam a conquista espacial para fazer propaganda
sobre os benefícios dos seus respectivos sistemas políticos.
Os soviéticos mandaram o primeiro homem a sobrevoar o espaço, o cosmonauta Yuri Gagarin.
Sentindo que ficavam para trás na corrida espacial, o presidente americano John Kennedy lança o desafio de pousar na Lua antes do final da década de 60."""

# ==============================================================================
# FUNÇÃO DE PRÉ-PROCESSAMENTO (EXERCÍCIO 1)
# ==============================================================================
def preprocess_text(text):
    """
    Converte o texto para minúsculas, remove números e pontuações,
    e retorna o texto limpo com quebras de linha preservadas.
    """
    # 1. Converter para minúsculas
    text_lower = text.lower()

    # 2. Remover números (usando expressão regular)
    text_no_numbers = re.sub(r'\d+', '', text_lower)

    # 3. Remover pontuações (usando a lista de pontuações do módulo string)
    translator = str.maketrans('', '', string.punctuation)
    text_no_punctuation = text_no_numbers.translate(translator)

    # 4. Limpeza e formatação, preservando quebras de linha
    # Substitui múltiplos espaços/tabulações por um único espaço
    text_cleaned = re.sub(r'[ \t]+', ' ', text_no_punctuation)
    # Remove espaços no início e fim de cada linha (quebras de linha)
    text_cleaned = re.sub(r' \n', '\n', text_cleaned)
    text_cleaned = re.sub(r'\n ', '\n', text_cleaned)
    text_cleaned = text_cleaned.strip()

    return text_lower, text_cleaned

# ==============================================================================
# FUNÇÃO PRINCIPAL DE ANÁLISE (EXERCÍCIO 2)
# ==============================================================================
def analyze_and_answer():

    # Pré-processamento para obter as versões do texto
    text_lower, text_cleaned = preprocess_text(TEXTO_ORIGINAL)

    # --------------------------------------------------------------------------
    # EXERCÍCIO 1: Exibir o texto original e o pré-processado
    # --------------------------------------------------------------------------
    print("="*50)
    print("EXERCÍCIO 1: PRÉ-PROCESSAMENTO DE TEXTO")
    print("="*50)

    print("\n" + "="*50)
    print("TEXTO ORIGINAL")
    print("="*50)
    print(TEXTO_ORIGINAL)

    print("\n" + "="*50)
    print("RESULTADO DO PRÉ-PROCESSAMENTO (Minúsculas, Sem Números e Pontuações)")
    print("="*50)
    print(text_cleaned)

    # --------------------------------------------------------------------------
    # EXERCÍCIO 2: Respostas
    # --------------------------------------------------------------------------
    print("\n\n" + "="*50)
    print("EXERCÍCIO 2: ANÁLISE DO TEXTO")
    print("="*50)

    # 1) Quantos caracteres tem o texto original?
    count_original = len(TEXTO_ORIGINAL)
    print(f"1) Quantos caracteres tem o texto original? R: {count_original}")

    # 2) Quantos caracteres tem o texto com as letras todas em formato minúsculo?
    count_lower = len(text_lower)
    print(f"2) Quantos caracteres tem o texto com as letras todas em formato minúsculo? R: {count_lower}")

    # 3) Quantos caracteres tem o texto sem números e sem pontuações?
    count_cleaned = len(text_cleaned)
    print(f"3) Quantos caracteres tem o texto sem números e sem pontuações? R: {count_cleaned}")

    # 4) Como saber se a palavra “Estados Unidos” aparece no texto original? E no texto transformado para letras minúsculas, como seria? Informe o código necessário em Python para os dois casos.
    print("\n4) Como saber se a palavra 'Estados Unidos' aparece no texto original? E no texto transformado para letras minúsculas, como seria? Informe o código necessário em Python para os dois casos.")

    # Verificação no Texto Original (Sensível a maiúsculas/minúsculas)
    keyword_original = "Estados Unidos"
    is_present_original = keyword_original in TEXTO_ORIGINAL

    print("\n--- Verificação no Texto Original (Sensível a maiúsculas/minúsculas) ---")
    print(f"Código: '{keyword_original}' in TEXTO_ORIGINAL")
    print(f"Resultado: {is_present_original}")

    # Verificação no Texto em Minúsculas (Insensível a maiúsculas/minúsculas)
    keyword_lower = "estados unidos"
    is_present_lower = keyword_lower in text_lower

    print("\n--- Verificação no Texto em Minúsculas (Insensível a maiúsculas/minúsculas) ---")
    print(f"Código: '{keyword_lower}' in text_lower")
    print(f"Resultado: {is_present_lower}")

if __name__ == "__main__":
    analyze_and_answer()
```
