```python
#2. Elaborar um novo código em Python capaz de calcular a média de ranking de um país ao longo dos anos, usando um gráfico de barras ou linhas.

import pandas as pd
import matplotlib.pyplot as plt

file_path = "fifa_ranking-2024-04-04.csv"
df = pd.read_csv(file_path)

df['rank_date'] = pd.to_datetime(df['rank_date'])
df_brasil = df[df['country_full'] == 'Brazil']

df_brasil['year'] = df_brasil['rank_date'].dt.year
media_ranking_brasil = df_brasil.groupby('year')['rank'].mean()

plt.figure(figsize=(12, 6))
plt.plot(media_ranking_brasil.index, media_ranking_brasil.values, marker='o', linestyle='-', color='blue', label='Média do Ranking')
plt.title("Média do Ranking do Brasil (1992-2024)", fontsize=16)
plt.xlabel("Ano", fontsize=12)
plt.ylabel("Média do Ranking", fontsize=12)
plt.gca().invert_yaxis()
plt.grid(axis="both", linestyle="--", alpha=0.7)
plt.legend()
plt.tight_layout()

plt.show()
```
