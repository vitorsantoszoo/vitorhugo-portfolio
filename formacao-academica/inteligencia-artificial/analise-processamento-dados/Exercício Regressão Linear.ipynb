```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Dados fornecidos
horas_esporte = np.array([0.5, 1.0, 1.5, 2.0, 3.0, 4.0]).reshape(-1, 1)  # Horas praticando esportes (X)
humor = np.array([3, 4, 5, 6, 7, 9])  # Classificação de humor (Y)

# Criar o modelo de regressão linear
modelo = LinearRegression()
modelo.fit(horas_esporte, humor)

# Coeficientes da reta
coeficiente_angular = modelo.coef_[0]  # Inclinação da reta
intercepto = modelo.intercept_  # Intercepto (onde cruza o eixo Y)
print(f"Equação da reta: Y = {coeficiente_angular:.2f}X + {intercepto:.2f}")

# Previsão para os valores de X
humor_previsto = modelo.predict(horas_esporte)

# Visualização dos dados
plt.scatter(horas_esporte, humor, color='blue', label='Dados reais')  # Pontos reais
plt.plot(horas_esporte, humor_previsto, color='red', label='Reta ajustada')  # Reta ajustada
plt.title('Regressão Linear Simples')
plt.xlabel('Horas praticando esportes')
plt.ylabel('Classificação de humor')
plt.legend()
plt.grid()
plt.show()
```
